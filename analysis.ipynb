```python
import os

def count_images_and_folders(folder_path):
    image_count = 0
    folder_count = 0
    for patient_folder in os.listdir(folder_path):
        patient_path = os.path.join(folder_path, patient_folder)
        if os.path.isdir(patient_path):
            folder_count += 1
            for file in os.listdir(patient_path):
                if file.lower().endswith(('.jpg', '.jpeg', '.png', '.tif')):
                    image_count += 1
    return image_count, folder_count

data_dir = "./data"
train_dir = os.path.join(data_dir, "train")
test_dir = os.path.join(data_dir, "test")

train_image_count, train_folder_count = count_images_and_folders(train_dir)
test_image_count, test_folder_count = count_images_and_folders(test_dir)

print(f"Train: {train_folder_count} folders, {train_image_count} images")
print(f"Test: {test_folder_count} folders, {test_image_count} images")
```

```python
import os
import pandas as pd

def get_image_counts(folder_path):
    """Returns a list of (patient_folder, image_count) tuples"""
    data = []
    for patient_folder in sorted(os.listdir(folder_path)):
        patient_path = os.path.join(folder_path, patient_folder)
        if os.path.isdir(patient_path):
            count = sum(
                1 for f in os.listdir(patient_path)
                if f.lower().endswith(('.jpg', '.jpeg', '.png', '.tif'))
            )
            data.append((patient_folder, count))
    return data

# Define paths
data_dir = "./data"  # <-- Adjust as needed
train_dir = os.path.join(data_dir, "train")
test_dir = os.path.join(data_dir, "test")

# Get image counts
train_data = get_image_counts(train_dir)
test_data = get_image_counts(test_dir)

# Convert to DataFrame
train_df = pd.DataFrame(train_data, columns=["Patient ID", "Image Count"])
test_df = pd.DataFrame(test_data, columns=["Patient ID", "Image Count"])

# Save to Excel with two sheets: Train and Test
output_file = "image_counts_by_patient.xlsx"
with pd.ExcelWriter(output_file) as writer:
    train_df.to_excel(writer, sheet_name="Train", index=False)
    test_df.to_excel(writer, sheet_name="Test", index=False)

print(f"Excel file saved as: {output_file}")
```

```python
import os
import pandas as pd

def count_images_and_folders(folder_path):
    image_count = 0
    folder_count = 0
    for patient_folder in os.listdir(folder_path):
        patient_path = os.path.join(folder_path, patient_folder)
        if os.path.isdir(patient_path):
            folder_count += 1
            for file in os.listdir(patient_path):
                if file.lower().endswith(('.jpg', '.jpeg', '.png', '.tif')):
                    image_count += 1
    return image_count, folder_count

# Paths
data_dir = "./data"
train_dir = os.path.join(data_dir, "train")
test_dir = os.path.join(data_dir, "test")

# Get counts
train_images, train_patients = count_images_and_folders(train_dir)
test_images, test_patients = count_images_and_folders(test_dir)

# Build summary DataFrame
summary = pd.DataFrame({
    "Split": ["Train", "Test", "Total"],
    "# Patients": [train_patients, test_patients, train_patients + test_patients],
    "# Images": [train_images, test_images, train_images + test_images]
})

# Print summary as table
print(summary.to_string(index=False))
```
